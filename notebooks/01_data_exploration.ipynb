```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

```python
import sys
print(sys.executable)
```

```python
TICKERS = ["SPY", "EFA", "AGG", "GLD", "DBC", "VNQ", "BIL"]
prices = yf.download(TICKERS, start="2000-01-01", auto_adjust=True)
prices.head()
```

```python
prices = prices["Close"]
prices.head()
```

```python
prices.notna().idxmax()
```

```python
for ticker in prices.columns:
    first_valid = prices[ticker].first_valid_index()
    series_after_start = prices[ticker].loc[first_valid:]
    nan_count = series_after_start.isna().sum()
    print(f"{ticker}: {nan_count} missing days after {first_valid.date()}")
```

```python
# Make sure the output directory exists
output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

# Save to Parquet
output_path = output_dir / "prices.parquet"
prices.to_parquet(output_path)

print(f"Saved {prices.shape[0]} rows × {prices.shape[1]} columns to {output_path}")
```

```python
prices_check = pd.read_parquet(output_path)
print(f"Loaded {prices_check.shape[0]} rows × {prices_check.shape[1]} columns")
print(f"Matches original: {prices.equals(prices_check)}")
```
